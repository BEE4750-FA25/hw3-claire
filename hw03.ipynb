### Problem 1.1

Implement the Streeter-Phelps (analytic) solution for the dissolved
oxygen concentration. Plot the dissolved oxygen concentration from the
first waste stream to 50 km downriver. What is the minimum value in
mg/L?

```julia
x2 = 15 # km
xf = 50 # km
dx = 0.1 # km
u = 6 # km/d

C_sat = 10 # mg/L

kr = 0.55 # day^-1
kc = 0.35 # day^-1
kn = 0.25 # day^-1

Qr = 100000 # m^3/day
Cdo_r = 7.5 # mg/L
Cc_r = 5 # mg/L
Cn_r = 5 # mg/L

Q1 = 10000 # m^3/day
Cdo_1 = 5 # mg/L
Cc_1 = 50 # mg/L
Cn_1 = 35 # mg/L

Q2 = 15000 # m^3/day
Cdo_2 = 5 # mg/L
Cc_2 = 45 # mg/L
Cn_2 = 35 # mg/L


# Mixing equation
C_new(q1, c1, q2, c2) = (q1 * c1 + q2 * c2) / (q1 + q2)

# Streeter-Phelps DO function
function mix(D0_0, CBOD0, NBOD0, t, k1, k2, k3, sat)
    D0_new = D0_0 * exp(-k1 * t)
    CBOD_new = CBOD0 * (k2 / (k1 - k2)) * (exp(-k2 * t) - exp(-k1 * t))
    NBOD_new = NBOD0 * (k3 / (k1 - k3)) * (exp(-k3 * t) - exp(-k1 * t))
    D = D0_new + CBOD_new + NBOD_new
    return sat - D   # convert deficit to actual DO
end

# Initial mixing at upstream + first waste
Q_first = Qr + Q1
DO_mix1 = C_new(Qr, Cdo_r, Q1, Cdo_1)
CBOD_mix1 = C_new(Qr, Cc_r, Q1, Cc_1)
NBOD_mix1 = C_new(Qr, Cn_r, Q1, Cn_1)
D0_first = C_sat - DO_mix1

decay(C_0, k, t) = C_0 * exp(-k * t)

xs = 0.0:dx:xf
DOs = Float64[]

# Initial point (x=0)
push!(DOs, mix(D0_first, CBOD_mix1, NBOD_mix1, 0.0, kr, kc, kn, C_sat))

for x in xs
    if x <= x2
        t = x / u
        push!(DOs, mix(D0_first, CBOD_mix1, NBOD_mix1, t, kr, kc, kn, C_sat))
    else
        # second discharge at 15 km
        t_x2 = x2 / u
        DO_x2 = mix(D0_first, CBOD_mix1, NBOD_mix1, t_x2, kr, kc, kn, C_sat)

        CBOD_x2 = decay(CBOD_mix1, kc, t_x2)
        NBOD_x2 = decay(NBOD_mix1, kn, t_x2)

        Q_second = Q_first + Q2
        DO_mix2 = C_new(Q_first, DO_x2, Q2, Cdo_2)
        CBOD_mix2 = C_new(Q_first, CBOD_x2, Q2, Cc_2)
        NBOD_mix2 = C_new(Q_first, NBOD_x2, Q2, Cn_2)
        D0_second = C_sat - DO_mix2

        t_after2 = (x - x2) / u
        push!(DOs, mix(D0_second, CBOD_mix2, NBOD_mix2, t_after2, kr, kc, kn, C_sat))
    end
end

# Plot
plot(xs, DOs,
    xlabel = "Distance downstream (km)",
    ylabel = "Dissolved oxygen (mg/L)",
    title = "Streeter–Phelps DO (two discharges: at 0 and $x2 km)",
    legend = false)
```
